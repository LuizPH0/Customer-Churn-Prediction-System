# Sprint 7 - Dashboard Dataset

## Etapa 1 - Carregar dataset

```python
import pandas as pd
```

```python
df = pd.read_csv(
    '../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv'
)
```

---

## Etapa 2 - Corrigir Total Charges

```python
df['TotalCharges'] = pd.to_numeric(
    df['TotalCharges'],
    errors='coerce'
)
```

---

## Etapa 3 - Remover missing values

```python
df.dropna(inplace=True)
```

## Etapa 4 - Salvar Dataset

```python
dashboard_df = df.copy()
```

```python
dashboard_df.to_csv(
    '../data/processed/dashboard_dataset.csv',
    index=False
)
```
