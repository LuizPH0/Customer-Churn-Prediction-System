# Sprint 6 - Feature Importance (Importância da funcionalidade)

## Etapa 1 - Importar bibliotecas

```python
import pandas as pd
import joblib
import matplotlib.pyplot as plt
```

---

## Etapa 2 - Carregar Modelo e Features

```python
#Carregar Modelo

model_lr = joblib.load(
    '../models/model_lr.pkl'
)
```

```python
#Carregar Features

feature_names = joblib.load(
    '../models/feature_names.pkl'
)
```

---

## Etapa 3 - Criar feature importance

```python
feature_importance = pd.DataFrame({
    'Feature': feature_names,
    'Coefficient': model_lr.coef_[0]
})
```

```python
feature_importance = feature_importance.sort_values(
    by='Coefficient',
    ascending=False
)
```

```python
#Features que mais aumentam o Churn

feature_importance.head(10)
```

```python
#As que mais reduzem o Churn

feature_importance.tail(10)
```

---

## Etapa 4 - Criar gráfico de análise

```python
#Top de 10 que mais aumentam Churn

top_positive = feature_importance.head(10)

plt.figure(figsize=(10,6))

plt.barh(
    top_positive['Feature'],
    top_positive['Coefficient']
)

plt.xlabel('Impacto no Churn')
plt.title('Top Features que Aumentam o Churn')
plt.gca().invert_yaxis()

plt.show()
```

```python
#Top 10 que mais diminuem o Churn

top_positive = feature_importance.tail(10)

plt.figure(figsize=(10,6))

plt.barh(
    top_positive['Feature'],
    top_positive['Coefficient']
)

plt.xlabel('Impacto no Churn')
plt.title('Top Features que Diminuem o Churn')


plt.show()
```

## Insights

Pessoas com serviço de internet fibra óptica, que pagam mais, que assinam o premium que inclui Streamings tem mais chance de cancelar.
A insatisfação pode estar associado ao serviço de fibra óptica e aos preços dos assinantes premiums.

Pessoas com estabilidade(tenure), que pagam mensalmente e pessoas com contratos maiores, online security, tech support e que tenham dependentes tem menores chances de cancelar.
Isso pode estar associado a assinantes que gozam do tech support e tem dependentes que usam o serviço diariamente.
