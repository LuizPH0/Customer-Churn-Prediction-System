# SPRINT 5 - Modelagem

### Objetivo:

Comparar modelos e descobrir qual prevê churn melhor.

Vamos começar com Logistic Regression. Por quê? Porque é baseline clássico de churn.

---

## Etapa 1 - Carregar Dataset Processado

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/processed/encoded_df.csv')

df.head()
```

---

## Etapa 2 — Conferir dataset

```python
df.shape
```

```python
df.columns
```

```python
df['Churn'].value_counts()
```

---

## Etapa 3 — Separar Features e Target

```python
#O X representa as variáveis independentes
X = df.drop('Churn', axis=1)

#O Y representa o target (churn)
y = df['Churn']
```

```python
print(X.shape)
print(y.shape)
```

---

## Etapa 4 — Dividir treino e teste

```python
#Importando biblioteca para treino

from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2, # 20% destinado a teste
    random_state=42,
    stratify=y
)
```

```python
print(X_train.shape)
print(X_test.shape)
```

---

## Etapa 5 — Escalonamento

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
```

```python
X_test_scaled = scaler.transform(X_test)
```

---

## Etapa 6 — Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
```

```python
#Criação do Modelo

model_lr = LogisticRegression(
    max_iter=1000,
    random_state=42
)
```

```python
#Treinamento

model_lr.fit(X_train_scaled, y_train)
```

---

## Etapa 7 — Predição

```python
y_pred_lr = model_lr.predict(X_test_scaled)
```

---

## Etapa 8 — Avaliação do modelo

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score
)
```

```python
#Acurácia

accuracy = accuracy_score(y_test, y_pred_lr) * 100

print(f'Accuracy: {accuracy:.2f}%')
```

```python
#F1-Score

score = f1_score(y_test, y_pred_lr) * 100
print(f'F1 Score: {score:.2f}%')
```

```python
#Classification Report

print(classification_report(y_test, y_pred_lr))
```

```python
#Matriz de Confusão

confusion_matrix(y_test, y_pred_lr)
```

---

## Conclusão do Modelo de Regressão Logística

A Regressão Logística apresentou desempenho inicial satisfatório como baseline, alcançando aproximadamente 80% de accuracy. No entanto, observou-se recall de 57% para a classe churn, indicando oportunidade de melhoria na capacidade do modelo em identificar clientes propensos ao cancelamento. Como o custo de falsos negativos é alto nesse problema de negócio, modelos adicionais serão comparados visando aumentar o recall

# Sprint 5.2 - Testando novos modelos

---

## Modelo 1 - Decision Tree (Árvore de Decisão)

```python
#Importar
from sklearn.tree import DecisionTreeClassifier
```

```python
#Criar o Modelo

model_dt = DecisionTreeClassifier(
    random_state=42,
    max_depth=5 # usado para não ocorrer overfitting
)
```

```python
#Treinar Modelo

model_dt.fit(X_train, y_train)
```

```python
#Predição

y_pred_dt = model_dt.predict(X_test)
```

```python
#Avaliação

print('Accuracy:',
      accuracy_score(y_test, y_pred_dt))

print(classification_report(
    y_test,
    y_pred_dt
))
print("Matriz de Confusão")
print(confusion_matrix(
    y_test,
    y_pred_dt
))
```

---

## Modelo 2 - Random Forest

```python
#Importar lib

from sklearn.ensemble import RandomForestClassifier
```

```python
#Criar Modelo

model_rf = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    max_depth=8
)

# Com 200 árvores para mais robustez e max_depth=8 para controlar o overfitting
```

```python
#Treinar

model_rf.fit(X_train, y_train)
```

```python
#Predição

y_pred_rf = model_rf.predict(X_test)
```

```python
#Avaliação

print('Accuracy:',
      accuracy_score(y_test, y_pred_rf))

print(classification_report(
    y_test,
    y_pred_rf
))
print("Matriz de Confusão")
print(confusion_matrix(
    y_test,
    y_pred_rf
))
```

---

Embora a Logistic Regression tenha apresentado melhor equilíbrio geral entre precisão e F1-score, em um cenário real de churn eu priorizaria recall, pois o custo de falsos negativos é elevado — clientes podem cancelar sem que a empresa tenha oportunidade de agir preventivamente

# Sprint 5.3 - Threshold Tuning

---

## Etapa 1 - Obter probabilidades

```python
y_prob_lr = model_lr.predict_proba(X_test_scaled)[:, 1]

y_prob_lr
```

---

## Etapa 2 — Testar thresholds

```python
#Importar a lib

from sklearn.metrics import classification_report
```

```python
#testar com 0.50

threshold = 0.50

y_pred_custom = (
    y_prob_lr >= threshold
).astype(int)

print(classification_report(
    y_test,
    y_pred_custom
))

print(confusion_matrix(
    y_test,
    y_pred_custom
))
```

```python
#testar com 0.40

threshold = 0.40

y_pred_custom = (
    y_prob_lr >= threshold
).astype(int)

print(classification_report(
    y_test,
    y_pred_custom
))

print(confusion_matrix(
    y_test,
    y_pred_custom
))
```

```python
#testar com 0.35

threshold = 0.35

y_pred_custom = (
    y_prob_lr >= threshold
).astype(int)

print(classification_report(
    y_test,
    y_pred_custom
))

print(confusion_matrix(
    y_test,
    y_pred_custom
))
```

```python
#Testar com 0.30

threshold = 0.30

y_pred_custom = (
    y_prob_lr >= threshold
).astype(int)

print(classification_report(
    y_test,
    y_pred_custom
))

print(confusion_matrix(
    y_test,
    y_pred_custom
))
```

---

| Threshold | Precision | Recall | F1 |
| --------- | --------: | -----: | -: |
| 0.50      |      0.65 |   0.57 |0.61|
| 0.40      |      0.58 |   0.68 |0.63|
| 0.35      |      0.54 |   0.72 |0.62|
| 0.30      |      0.51 |   0.76 |0.61|

Foram avaliados diferentes thresholds de decisão para otimizar a identificação de clientes com risco de churn. O threshold padrão (0.50) apresentou recall limitado. Após ajuste, o threshold 0.40 demonstrou melhor equilíbrio entre precisão e capacidade preditiva, elevando o recall de 57% para 68% sem perda excessiva de precisão.
