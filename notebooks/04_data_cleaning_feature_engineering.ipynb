## Etapa 1 - Importar bibliotecas

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Configuração visual
pd.set_option('display.max_columns', None)
```

---

## Etapa 2 - Carregar dados

```python
df = pd.read_csv('../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv')

df.info()
```

---

## Etapa 3 - Corrigir Dados

```python
df['TotalCharges'].head(10)
```

```python
## Converter TotalCharges em número (float)

df['TotalCharges'] = pd.to_numeric(
    df['TotalCharges'],
    errors='coerce'
)
```

```python
#Verificar dados faltantes

df.isnull().sum()
```

```python
#Tratar missing values

df.dropna(inplace=True)

#Veriifcar se há dados faltantes ainda
df.isnull().sum()
```

---

## Etapa 4 - Preparar Dataset ML

```python
#Fazendo uma cópia para o dataset ML e dropando a coluna customerID que será desnecessária para o treinamento do modelo.

df_ml = df.copy()

df_ml.drop('customerID', axis=1, inplace=True)
```

---

## Etapa 5 — Transformar target (Churn)

```python
df_ml['Churn'] = df_ml['Churn'].map({
    'Yes': 1,
    'No': 0
})
```

```python
df_ml['Churn'].value_counts()
```

---

## Etapa 6 - Verificar colunas categóricas

```python
#Verificar quais colunas são texto, porque o ML não entende texto.

df_ml.select_dtypes(include='object').columns
```

---

## Etapa 7 - Encoding

```python
#Utilizando o 'get_dummies' porque é mais simples e fácil.

df_ml_encoded = pd.get_dummies(
    df_ml,
    drop_first=True
)
```

```python
df_ml_encoded.head()

df_ml_encoded.to_csv('encoded_df.csv', index=False)
```

---

## Etapa 8 - Separar X e Y

```python
#Features

X = df_ml_encoded.drop('Churn', axis=1)

#Target

y = df_ml_encoded['Churn']
```

```python
print(X.shape)
print(y.shape)
```

---

## Etapa 9 - Treino/Test Split

```python
#Importar o 'train_test_split'

from sklearn.model_selection import train_test_split
```

```python
#Treino

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2, #Usando 20% para teste
    random_state=42,
    stratify=y
)
```
